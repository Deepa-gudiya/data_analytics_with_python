BAR PLOT


A Bar Plot is a data visualization technique used to represent **categorical data** with rectangular bars.

Each bar represents a category, and the height (or length) of the bar represents the value corresponding to that category.

*When to Use a Bar Plot?

Bar plots are used when:

- Comparing values across different categories
- Showing differences between groups
- Visualizing discrete or categorical data
- Displaying frequency counts

*Real-World Examples

Bar plots are commonly used to visualize:

- Student marks in different subjects
- Sales comparison across different products
- Employee count by department
- Survey results by category
- Monthly revenue comparison

*Why Bar Plots Are Important?

- Easy to interpret
- Clear visual comparison
- Effective for presentations
- Helps identify highest and lowest values quickly

In This Example:
The x-axis represents discrete categories.

The y-axis represents corresponding numerical values.  

Using Matplotlib, we generate a simple bar plot to understand how categorical comparison works.


```python
import matplotlib.pyplot as plt
x=[1,5,6,7,3]
y=[2,3,6,9,4]
plt.bar(x,y)
plt.show()
```

```python
#ALIGN="EDGE"
import matplotlib.pyplot as plt
x=["python","c","c++","java","oracle"]
y=[34,44,64,54,24]
plt.xlabel("Language",fontsize=14)
plt.ylabel("Number of students")
plt.title("Course Enrollments",fontsize=15)
plt.bar(x,y,width=0.4,color=["red", "yellow", "blue", "orange", "violet"] ,align="edge")
plt.show
```

```python
#EDGE COLOR & LINE WIDTH
plt.bar(x,y,width=0.4,color=["red", "yellow", "blue", "orange", "violet"] ,edgecolor="k",linewidth=5)
```

```python
#ALPHA controls transparency level of the plot
plt.bar(x,y,width=0.4,color=["red", "yellow", "blue", "orange", "violet"] ,edgecolor="k",alpha=0.5)
```

```python
#LEGEND tells what each color/line represents
import matplotlib.pyplot as plt
x=["python","c","c++","java","oracle"]
y=[34,44,64,54,24]
plt.xlabel("Language",fontsize=14)
plt.ylabel("Number of students")
plt.title("Course Enrollments",fontsize=15)
plt.bar(x,y,width=0.4,color=[ "blue" ] ,edgecolor="k",
        label="popularity")
plt.legend()
plt.show()
```

```python
#DOUBLE BAR GRAPH-OVERLAPPED
import matplotlib.pyplot as plt
x=["python","c","c++","java","oracle"]
y=[34,44,64,54,24]
z=[20,30,50,30,10]
plt.xlabel("Language",fontsize=14)
plt.ylabel("Number of students")
plt.title("Course Enrollments",fontsize=15)
plt.bar(x,y,width=0.4,color=[ "blue" ] ,edgecolor="k",
        label="popularity")
plt.bar(x,z,width=0.4,color=[ "y" ] ,edgecolor="k",
        label="popularity1")
plt.legend()
plt.show()
```

```python
#DOUBLE BAR GRAPH -followed by side
import matplotlib.pyplot as plt
import numpy as np
x=["python","c","c++","java","oracle"]
y=[34,44,64,54,24]
z=[20,30,50,30,10]
width=0.2
p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel("Language",fontsize=14)
plt.ylabel("Number of students")
plt.title("Course Enrollments",fontsize=15)

plt.bar(p,y,width,color=[ "blue" ] ,edgecolor="k",
        label="popularity")
plt.bar(p1,z,width,color=[ "y" ] ,edgecolor="k",
        label="popularity1")
plt.legend()
plt.show()
```

```python
#x-label with names "XTICKS"
import matplotlib.pyplot as plt
import numpy as np
x=["python","c","c++","java","oracle"]
y=[34,44,64,54,24]
z=[20,30,50,30,10]
width=0.2
p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel("Language",fontsize=14)
plt.ylabel("Number of students")
plt.title("Course Enrollments",fontsize=15)

plt.bar(p,y,width,color=[ "blue" ] ,edgecolor="k",
        label="popularity")
plt.bar(p1,z,width,color=[ "y" ] ,edgecolor="k",
        label="popularity1")

plt.xticks(p+width,x) #  or P+width/2
plt.xticks(p+width,x,rotation=20) #ROTATION
plt.legend()
plt.show()
```

```python
#HORIZONTAL BAR GRAPH 
import matplotlib.pyplot as plt
x=["python","c","c++","java","oracle"]
y=[34,44,64,54,24]
z=[20,30,50,30,10]

plt.xlabel("Language",fontsize=14)
plt.ylabel("Number of students")
plt.title("Course Enrollments",fontsize=15)

plt.barh(x,y,color=[ "blue" ] ,edgecolor="k",
        label="popularity")
plt.bar(x,z,color=[ "y" ] ,edgecolor="k",
        label="popularity1")
```
