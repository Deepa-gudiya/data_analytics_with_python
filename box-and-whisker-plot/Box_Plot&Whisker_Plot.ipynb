📦 Box Plot (Box-and-Whisker Plot) 

A Box Plot, also known as a Box-and-Whisker Plot, is a statistical visualization used to summarize the distribution of a dataset.

It provides a clear graphical representation of:

Minimum value (excluding outliers)

First Quartile (Q1 – 25th percentile)

Median (Q2 – 50th percentile)

Third Quartile (Q3 – 75th percentile)

Maximum value (excluding outliers)

Outliers

The “box” represents the Interquartile Range (IQR = Q3 − Q1), while the “whiskers” extend to show the spread of the data. Any points beyond the whiskers are considered outliers.

🎯 Why Use a Box Plot?

To understand data spread and variability

To detect outliers

To compare distributions across multiple categories

To quickly summarize large datasets

📊 Real-World Applications

Salary distribution analysis

Student marks comparison

Product price variation

Delivery time analysis

Performance evaluation metrics

Import Required Libraries

```python
# Import matplotlib for visualization
import matplotlib.pyplot as plt

# Import numpy for numerical operations
import numpy as np
```

Basic Box Plot

-The box represents the Interquartile Range (Q1 to Q3).

-The line inside the box represents the Median.

-The whiskers extend to minimum and maximum values.

-100 is detected as an outlier.

```python
# Create a sample dataset
data = np.array([10, 15, 20, 25, 30, 35, 40, 100])

# Create a basic box plot
plt.boxplot(data)

# Add title and label
plt.title("Basic Box Plot")
plt.ylabel("Values")

# Display the plot
plt.show()
```

Orientation (vert) 

vert=False changes orientation from vertical to horizontal.

```python
plt.boxplot(data, vert=False)  # Horizontal box plot
plt.title("Horizontal Box Plot")
plt.show()
```

Fill Color (patch_artist)

Allows customization of box color for professional styling.

```python
plt.boxplot(data, patch_artist=True)  # Enables color filling
plt.title("Filled Box Plot")
plt.show()
```

Show Mean Value (showmeans)

Adds a marker to show the mean value along with median.

```python
plt.boxplot(data, showmeans=True)  # Displays mean marker
plt.title("Box Plot with Mean")
plt.show()
```

Control Outliers (showfliers)

Removes visual display of outliers.

```python
plt.boxplot(data, showfliers=False)  # Hides outliers
plt.title("Box Plot without Outliers")
plt.show()
```

Notched Box Plot (notch)

Displays confidence interval around median, useful for comparison.

```python
plt.boxplot(data, notch=True)  # Shows confidence interval around median
plt.title("Notched Box Plot")
plt.show()
```

Whisker Length Control (whis)

Defines how far whiskers extend (based on IQR multiplier).

```python
plt.boxplot(data, whis=1.0)  # Controls whisker length
plt.title("Custom Whisker Length")
plt.show()
```

Multiple Dataset Comparison

Allows comparison of distribution between multiple groups.

```python
# Create multiple datasets
data1 = [10, 20, 30, 40, 50]
data2 = [15, 25, 35, 45, 55]

# Compare multiple datasets
plt.boxplot([data1, data2], labels=["Team A", "Team B"])

plt.title("Multiple Dataset Comparison")
plt.ylabel("Scores")

plt.show()
```

meanline

Displays the mean as a line instead of a point.

```python
plt.boxplot(data, showmeans=True, meanline=True)
```

medianprops

Controls color and thickness of median line.

```python
plt.boxplot(data, medianprops=dict(color='red', linewidth=2))
```

boxprops

Controls box border styling.

```python
plt.boxplot(data, boxprops=dict(color='blue', linewidth=2))
```

whiskerprops

Changes whisker color/style.

```python
plt.boxplot(data, whiskerprops=dict(color='green'))
```

capprops

Styles the small horizontal lines at whisker ends.

```python
plt.boxplot(data, capprops=dict(color='black'))
```

flierprops

Controls outlier shape, size, and color.

```python
plt.boxplot(
    data,
    flierprops=dict(marker='o', markerfacecolor='red', markersize=8)
)
```

positions

Control box positions manually.

```python
plt.boxplot([data1, data2], positions=[1,3])
```

manage_ticks

Controls automatic tick management.

```python
plt.boxplot(data, manage_ticks=False)
```

sym

Change outlier symbol.


```python
plt.boxplot(data, sym='r+')
```

widths

Controls box width.

```python
plt.boxplot(data, widths=0.3)
```

bootstrap

Improves notch confidence interval

```python
plt.boxplot(data, notch=True, bootstrap=1000)
```

usermedians

Manually set median value.

```python
plt.boxplot(data, usermedians=[50])
```

conf_intervals

Manually set confidence intervals for notch.

```python
plt.boxplot(data, notch=True, conf_intervals=[[45, 55]])
```

zorder

Controls layering of plot elements.

Higher value → appears on top

```python
plt.boxplot(data, zorder=3)
```

autorange

Automatically adjusts whiskers if IQR is zero.

```python
plt.boxplot(data, autorange=True)
```

labels (Important for Multiple Data)

Adds category names on x-axis

```python
plt.boxplot([data1, data2], labels=['Class A', 'Class B'])
```

patch_artist

Allows filling the box with color.

Required if you want colored boxes.

```python
plt.boxplot(data, patch_artist=True)
```

📦 Real-World Scenario: Employee Salary Distribution Analysis

 Scenario:
A company wants to analyze the annual salary distribution 
across three departments:

- IT Department
- HR Department
- Sales Department

The objective is to:
- Compare salary spread
- Identify salary consistency
- Detect outliers (extremely high or low salaries)
- Compare median and mean salaries

A Box-and-Whisker Plot is used to summarize the distribution effectively.


Step 1: Import Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
```

Step 2: Create Sample Salary Data

```python
IT = [4, 5, 6, 7, 8, 9, 10, 12, 15]
HR = [3, 4, 4, 5, 6, 6, 7, 8, 9]
Sales = [2, 3, 4, 5, 6, 7, 8, 20]

data = [IT, HR, Sales]
```

Step 3: Box Plot Covering Major Parameters

```python
plt.figure()

plt.boxplot(
    data,
    
    # Structure Parameters
    vert=True,                      # Vertical orientation
    widths=0.5,                     # Width of boxes
    positions=[1, 2, 3],            # Custom positions
    patch_artist=True,              # Enable box coloring
    labels=["IT", "HR", "Sales"],   # Category labels
    
    # Statistical Parameters
    whis=1.5,                       # IQR whisker rule
    notch=True,                     # Show median confidence interval
    bootstrap=1000,                 # Improve notch accuracy
    
    # Display Parameters
    showmeans=True,                 # Show mean value
    meanline=True,                  # Mean shown as line
    showfliers=True,                # Show outliers
    autorange=True,                 # Auto adjust whiskers if needed
    
    # Styling Parameters
    boxprops=dict(color='blue', linewidth=2),
    medianprops=dict(color='red', linewidth=2),
    whiskerprops=dict(color='green', linestyle='--'),
    capprops=dict(color='black', linewidth=2),
    flierprops=dict(marker='o', markerfacecolor='orange', markersize=6),
    
    zorder=2                        # Layer control
)

plt.title("Employee Salary Distribution by Department")
plt.xlabel("Departments")
plt.ylabel("Salary (in Lakhs per annum)")

plt.show()
```

Explanation:

Interpretation

- The box represents the middle 50% of salaries (Interquartile Range).
- The red line inside the box indicates the median salary.
- The green dashed lines (whiskers) show salary spread within IQR limits.
- The black caps indicate the range boundary.
- Orange circles represent outliers (extremely high salaries).
- The mean salary is displayed as a horizontal line.
- Notches show the confidence interval around the median.

  Observations:
- IT department shows higher salary variation.
- Sales department contains an outlier (20 LPA).
- HR salaries are more consistent and compact.
