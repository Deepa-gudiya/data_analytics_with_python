 Pie Chart – Introduction

A Pie Chart is a circular data visualization technique used to represent **proportional or percentage-based data**.

Each slice of the pie represents a category, and the size of each slice corresponds to its contribution to the whole.

 When to Use a Pie Chart?

Pie charts are used when:

- Showing percentage distribution
- Comparing parts of a whole
- Visualizing proportional data
- Displaying category contribution

 Real-World Examples

Pie charts are commonly used to visualize:

- Market share of different companies
- Budget allocation by category
- Student distribution by grade
- Survey response percentages
- Expense breakdown (Rent, Food, Travel, etc.)

Why Pie Charts Are Important?

- Clearly show proportion of each category
- Easy to understand at a glance
- Effective for non-technical audiences
- Good for presentations and dashboards

Important Note:

Pie charts work best when:
- The number of categories is limited (ideally less than 6)
- The total represents 100%
- The goal is to show proportion, not precise comparison

    In This Example:

Each slice represents a category's contribution to the total dataset.

Using Matplotlib, we generate a pie chart to visualize how different categories contribute proportionally.


Import Required Library


```python
# Import matplotlib for visualization
import matplotlib.pyplot as plt
```

Basic Pie Chart

```python
# Sample categorical data (Product Sales)
'''sales → Numerical values
Each value becomes a slice of the circle
'''
products = ["Electronics", "Clothing", "Groceries", "Books"]
sales = [40, 25, 20, 15]

# Basic pie chart
plt.pie(sales)

# Display the chart
plt.show()
```

labels Parameter

```python
plt.pie(sales, labels=products)  #Adds category names to each slice.
plt.title("Basic Pie Chart with Labels")
plt.show()
```

autopct (Show Percentage)

```python
plt.pie(sales, labels=products, autopct='%1.1f%%')    #%1.1f%% → Shows percentage with 1 decimal
plt.title("Pie Chart with Percentages")
plt.show()
```

colors Parameter

```python
colors = ["gold", "lightblue", "lightgreen", "pink"]  #Improves visualization clarity.

plt.pie(sales, labels=products, colors=colors, autopct='%1.1f%%')
plt.title("Pie Chart with Custom Colors")
plt.show()
```

explode (Highlight Slice)

```python
explode = (0.1, 0, 0, 0)  #Used to emphasize important category.

plt.pie(sales, labels=products, autopct='%1.1f%%', explode=explode) 
plt.title("Pie Chart with Exploded Slice")
plt.show()
```

shadow Parameter

```python
plt.pie(sales, labels=products, autopct='%1.1f%%', shadow=True) #Adds depth effect.
plt.title("Pie Chart with Shadow Effect")
plt.show()
```

startangle (Rotate Chart)

```python
plt.pie(sales, labels=products, autopct='%1.1f%%', startangle=180) #Changes starting position of first slice.
plt.title("Pie Chart Rotated by 90 Degrees")
plt.show()
```

wedgeprops (Slice Border Styling)

```python
'''Adds professional borders around slices'''
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        wedgeprops={'edgecolor': 'black', 'linewidth': 2})

plt.title("Pie Chart with Border Styling")
plt.show()
```

textprops (Text Styling)

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        textprops={'fontsize': 15, 'color': 'violet'}) #Customize font size and color.

plt.title("Pie Chart with Styled Text")
plt.show()
```

counterclock Parameter

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        counterclock=False) #Controls direction of slices.

plt.title("Pie Chart in Clockwise Direction")
plt.show()
```

radius Parameter

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        radius=0.8)  #Controls size of pie chart.

plt.title("Pie Chart with Custom Radius")
plt.show()
```

frame Parameter

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        frame=True)

plt.title("Pie Chart with Frame")
plt.show()
```

axis('equal') 

```python
plt.pie(sales, labels=products, autopct='%1.1f%%')
plt.axis('equal')  # Makes circle perfectly round
plt.title("Perfect Circular Pie Chart") #Without this, pie may look oval.
plt.show()
```

Donut Chart

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        wedgeprops={'width': 0.4})  # Creates donut shape

plt.title("Donut Chart")
plt.axis('equal')
plt.show()
```

```python
#Complete Professional Version
products = ["Electronics", "Clothing", "Groceries", "Books"]
sales = [40, 25, 20, 15]

explode = (0.1, 0, 0, 0)

plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        colors=["gold", "lightblue", "lightgreen", "pink"],
        explode=explode,
        shadow=True,
        startangle=90,
        wedgeprops={'edgecolor': 'black'})

plt.title("Product Sales Distribution Analysis")
plt.axis('equal')
plt.show()
```

pctdistance
Controls distance of percentage text from center.

```python
import matplotlib.pyplot as plt

products = ["A", "B", "C"]
sales = [50, 30, 20]

plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        pctdistance=0.8)  # Move percentage outward

plt.axis('equal')
plt.title("Using pctdistance")
plt.show()
```

label distance-Controls distance of labels from center.

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        labeldistance=1.2) #Prevents overlapping labels.

plt.axis('equal')
plt.title("Using labeldistance")
plt.show()
```

center
Changes center position of pie.

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        center=(0.2, 0))

plt.axis('equal')
plt.title("Using center parameter")
plt.show()
```

rotatelabels
Rotates labels according to slice angle.

```python
plt.pie(sales,
        labels=products,
        autopct='%1.1f%%',
        rotatelabels=True)

plt.axis('equal')
plt.title("Using rotatelabels")
plt.show()
```

Multiple Pie Charts in One Figure

```python
fig, ax = plt.subplots(1, 2)

ax[0].pie(sales, labels=products, autopct='%1.1f%%')
ax[0].set_title("Pie Chart 1")

ax[1].pie([40, 35, 25], labels=["X", "Y", "Z"], autopct='%1.1f%%')
ax[1].set_title("Pie Chart 2")

plt.show()
```

Example: Monthly Expense Distribution Analysis

```python
# Import matplotlib for data visualization
import matplotlib.pyplot as plt
```

```python
# Monthly expense categories (realistic data)
categories = ["Rent", "Food", "Transport", "Entertainment", "Savings"]

# Amount spent in each category (in ₹)
expenses = [15000, 6000, 3000, 2000, 4000]
```

```python
# Create basic pie chart
plt.pie(expenses)

plt.title("Monthly Expense Distribution")
plt.show()
```

```python
plt.pie(expenses,
        labels=categories,          # Category names
        autopct='%1.1f%%')          # Show percentage with 1 decimal

plt.title("Monthly Expense Distribution")
plt.axis('equal')                   # Ensures circle shape

plt.show()
```

```python
plt.pie(expenses,
        labels=categories,          # Category names
        autopct='%1.1f%%')          # Show percentage with 1 decimal

plt.title("Monthly Expense Distribution")
plt.axis('equal')                   # Ensures circle shape

plt.show()
```

```python
# Highlight the largest expense (Rent)
explode = (0.1, 0, 0, 0, 0)

colors = ["gold", "lightgreen", "lightblue", "pink", "violet"]

plt.pie(expenses,
        labels=categories,
        autopct='%1.1f%%',
        explode=explode,           # Highlight Rent
        colors=colors,
        shadow=True,
        startangle=90,
        wedgeprops={'edgecolor': 'black'})

plt.title("Monthly Expense Distribution Analysis")
plt.axis('equal')

plt.show()
```
