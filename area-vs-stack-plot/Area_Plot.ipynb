 Area Plot (Using fill_between)

 Introduction

An Area Plot is a type of data visualization used to display 
the magnitude of values over a continuous interval (usually time).

It is created in Matplotlib using the `fill_between()` function.

Unlike a line plot, an area plot fills the region between 
a curve and a baseline, making it easier to visualize volume 
and magnitude.
    
    “An Area Plot created using fill_between() visualizes the magnitude between two boundaries, 
    making it useful for highlighting trends, thresholds, 
    and comparative regions in continuous data.”
    

 Why Use Area Plot?

- To show trends over time
- To emphasize magnitude
- To highlight differences between two curves
- To visualize threshold-based conditions


 Real-World Scenario

A company wants to visualize its monthly revenue growth 
throughout the year.

The area under the revenue curve helps management understand 
overall business performance and growth trend.

```python
import matplotlib.pyplot as plt
import numpy as np

```

```python
x = np.arange(1, 6)
y = [10, 20, 15, 25, 30]
```

```python
plt.fill_between(x, y)  #Defines X-axis values.
plt.title("Using x Parameter")
plt.show()
```

2.y1

Upper boundary of filled region.

```python
plt.fill_between(x, y)
plt.title("Using y1 Parameter")
plt.show()
```

3.y2

Lower boundary of fill (default = 0)

Useful when comparing two curves or filling between two lines.

```python
plt.fill_between(x, y, 5)
plt.title("Using y2 Parameter")
plt.show()
```

4. where
   
Fills area only where condition is True.

it is used To highlight specific regions (e.g., revenue above target).

```python
plt.fill_between(x, y, where=(np.array(y) > 15))
plt.title("Using where Parameter")
plt.show()
```

5.interpolate

Smoothly fills boundary where condition changes.
    

```python
plt.fill_between(x, y, where=(np.array(y) > 15), interpolate=True)
plt.title("Using interpolate Parameter")
plt.show()
```

6.step

Creates step-style area instead of smooth curve.

Used in discrete data like stock price changes.

```python
plt.fill_between(x, y, step='mid')
plt.title("Using step Parameter")
plt.show()
```

7.color

Sets fill color.

Improves readability and aesthetics

```python
plt.fill_between(x, y, color='green')
plt.title("Using color Parameter")
plt.show()
```

8.alpha

Controls transparency (0 = invisible, 1 = solid)

Useful when overlapping multiple areas.

```python
plt.fill_between(x, y, alpha=0.3)
plt.title("Using alpha Parameter")
plt.show()
```

9.label

Adds legend entry.

```python
plt.fill_between(x, y, label="Sales Data")
plt.legend()
plt.title("Using label Parameter")
plt.show()
```

10.edgecolor

Sets border color.

```python
plt.fill_between(x, y, edgecolor='black')
plt.title("Using edgecolor Parameter")
plt.show()
```

11.linewidth

Controls thickness of border line.

Highlights area boundaries.

```python
plt.fill_between(x, y, edgecolor='black', linewidth=3)
plt.title("Using linewidth Parameter")
plt.show()
```

12. hatch

Adds pattern inside filled region.

Useful in grayscale printing or reports.

```python
plt.fill_between(x, y, hatch='//')
plt.title("Using hatch Parameter")
plt.show()
```

13.zorder

Controls layering of plot elements.
    
When multiple plots overlap, higher zorder appears on top.


```python
plt.fill_between(x, y, zorder=3)
plt.title("Using zorder Parameter")
plt.show()
```

14.facecolor

Specifies fill color (same as color).

Gives more explicit control when combining edgecolor and facecolor separately

```python
plt.fill_between(x, y, facecolor='lightgreen')
```

15.linestyle
Changes border line style.

Useful when differentiating multiple areas.

'-' solid

'--' dashed

':' dotted

'-.' dash-dot
    

```python
plt.fill_between(x, y, edgecolor='black', linestyle='--')
```

16.antialiased

Smooths the edges of the filled area.

 Improves visual quality.
    

```python
plt.fill_between(x, y, antialiased=True)
```

17.data

Allows passing data using dictionary reference.

Useful when working with structured datasets or Pandas.

```python
plt.fill_between("x", "y", data={"x": x, "y": y})
```

18.rasterized

Converts plot to raster format inside vector graphics.

Improves performance for very large datasets in PDF exports.

```python
plt.fill_between(x, y, rasterized=True)
```

19.transform

Applies coordinate transformation.

Advanced plotting (rarely used in basic analytics).

```python
plt.fill_between(x, y, transform=plt.gca().transData)
```

 Real-World Example: Company Profit Analysis Using Area Plot

 🎯 Scenario

A company wants to analyze its monthly profit performance.

The management wants to:

- Visualize profit growth across months
- Highlight months where profit exceeded ₹50 lakhs
- Compare profit with a fixed minimum guarantee baseline
- Improve visualization clarity using styling and formatting
- Ensure professional presentation for reports

We will use an Area Plot (fill_between) to visualize this data.

Step 1: Import Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
```

Step 2: Create Real-World Data

```python
# Months (1–12)
months = np.arange(1, 13)

# Monthly profit (in Lakhs)
profit = np.array([20, 35, 40, 45, 55, 60, 50, 65, 70, 75, 80, 90])

# Minimum guaranteed profit baseline
baseline = 30
```

Step 3: Area Plot Covering All Parameters

```python
plt.figure(figsize=(10,6))

plt.fill_between(
    months,                    # x → X-axis values (months)
    profit,                    # y1 → Upper boundary (profit data)
    baseline,                  # y2 → Lower boundary (baseline profit)

    where=(profit > 50),       # Fill only where profit > 50
    interpolate=True,          # Smooth boundary when condition changes
    step=None,                 # Normal curve (options: 'pre', 'post', 'mid')

    color='skyblue',           # Fill color
    facecolor='lightblue',     # Interior color (same as color but explicit)
    alpha=0.6,                 # Transparency level

    edgecolor='black',         # Border color
    linewidth=2,               # Border thickness
    linestyle='-',             # Border style

    hatch='//',                # Pattern inside area
    label="Monthly Profit > 50 Lakhs",  # Legend label

    antialiased=True,          # Smooth edges
    zorder=2                   # Layer positioning
)

plt.title("Company Monthly Profit Analysis")
plt.xlabel("Month")
plt.ylabel("Profit (in Lakhs)")
plt.legend()

plt.show()
```

Explanation of Parameters Used

 1. x (months)
Defines horizontal axis values representing time.

2. y1 (profit)
Upper boundary of the filled area — actual profit values.

3. y2 (baseline)
Lower boundary — minimum guaranteed profit.

4. where
Fills only months where profit exceeds ₹50 Lakhs.

5. interpolate
Ensures smooth transition where condition changes.

6. step
Defines curve style (normal here). Can use 'pre', 'post', 'mid'.

7. color / facecolor
Controls fill color inside the area.

8. alpha
Controls transparency for better readability.

9. edgecolor
Defines border color.

10. linewidth
Controls thickness of border.

11. linestyle
Controls border style (solid, dashed, etc.).

12. hatch
Adds visual pattern inside area (useful for reports).

13. label
Used for legend display.

14. antialiased
Smooths plot edges.

15. zorder
Controls layering when multiple plots overlap.
