 Scatter Plot – Introduction

A Scatter Plot is used to visualize the **relationship between two numerical variables**.

Each point represents one observation in the dataset.

 When to Use a Scatter Plot?

- To check correlation between variables
- To identify trends
- To detect outliers
- To analyze patterns

 Real-World Examples

- Study hours vs Exam score
- Experience vs Salary
- Advertising spend vs Sales
- Height vs Weight

 Why Scatter Plots Are Important?

- Helps identify positive/negative correlation
- Useful in regression analysis
- Shows clusters and patterns clearly


SCATTER PLOT


```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y=[2,3,4,1,5,3,6]
plt.scatter(x,y)
plt.title("scatter plot",fontsize=15)
plt.show()
```

```python
#labels
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y=[2,3,4,1,5,3,6]
plt.scatter(x,y)
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
#colors
import matplotlib.pyplot as plt
day=[1,2,3,4,5,6,7]
no=[2,3,4,1,5,3,6]
colors=["r","y","g","b","r","g","r"]
plt.scatter(day,no,c=colors)
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
#sizes
import matplotlib.pyplot as plt
day=[1,2,3,4,5,6,7]
no=[2,3,4,1,5,3,6]
colors=["r","y","g","b","r","g","r"]
sizes=[10,20,40,50,60,50,30]
plt.scatter(day,no,c=colors,s=sizes)
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
#alpha
plt.scatter(day,no,c=colors,s=sizes,alpha=0.3)
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
#marker
plt.scatter(day,no,c=colors,s=sizes,marker="*")
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
#egde color-line width
plt.scatter(day,no,c=colors,s=sizes,marker="*",edgecolor="k",linewidth=0.5)
plt.title("scatter plot",fontsize=15)
plt.xlabel("Day",fontsize=15)
plt.ylabel("No.",fontsize=15)
plt.show()
```

```python
'''Colormaps help us visualize intensity, magnitude, or variation in data clearly.
c=colors → data values

cmap='viridis' → colormap

plt.colorbar() → shows color scale'''

import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, c=colors, cmap='viridis',alpha=0.5)
plt.colorbar()
plt.show()
```

 Scatter Plot – Study Hours vs Exam Scores

-- Objective
To analyze the relationship between Study Hours and Exam Scores.

-- Variables
- X-axis → Study Hours
- Y-axis → Exam Scores

Each point represents one student.

If the points move upward → Positive Correlation  
If downward → Negative Correlation  
If scattered randomly → No strong relationship

This visualization helps in understanding performance trends.


Step-1:Import Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
```

Step-2:Create Dataset

```python
study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
exam_scores = np.array([35, 40, 50, 55, 65, 70, 80, 90])
```

Step-3:Create Scatter Plot

```python
plt.figure(figsize=(10,6), dpi=100)

plt.scatter(
    study_hours,                 # X values
    exam_scores,                 # Y values
    s=150,                       # Marker size
    c=exam_scores,               # Color based on values
    cmap='viridis',              # Colormap
    marker='o',                  # Marker style
    alpha=0.8,                   # Transparency
    edgecolors='black',          # Border color
    linewidths=1.5,              # Border thickness
    label='Student Data'         # Legend label
)

plt.title(" Scatter Plot – Study Hours vs Exam Scores", fontsize=16, fontweight='bold')
plt.xlabel("Study Hours", fontsize=13)
plt.ylabel("Exam Scores", fontsize=13)

plt.xticks(np.arange(0, 10, 1))
plt.yticks(np.arange(30, 101, 10))

plt.xlim(0, 9)
plt.ylim(30, 100)

plt.grid(True, linestyle='--', alpha=0.6)

plt.colorbar(label='Exam Score Intensity')

plt.legend()
plt.tight_layout()
plt.show()
```
