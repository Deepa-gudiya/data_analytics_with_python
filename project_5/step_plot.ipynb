 Step Plot – Introduction

A Step Plot displays data points as steps instead of straight lines.

It shows sudden changes clearly.

When to Use Step Plot?

- When values change at specific intervals
- For discrete time changes
- For digital signals

Real-World Examples

- Stock price movement
- Electricity billing slabs
- Inventory level tracking

Import Libraries

```python
# Data visualization
import matplotlib.pyplot as plt

# Numerical operations
import numpy as np
```

 Basic Step Plot

```python
# Sample discrete interval data
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 15, 25, 30])

# Basic step plot
plt.step(x, y)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Basic Step Plot")

plt.show()
```

Important Parameters in Step Plot

1.where Parameter 

where='pre'

Value change occurs at beginning of interval

```python
plt.step(x, y, where='pre')
plt.title("Step Plot - Pre")
plt.show()
```

where='post'

Value change occurs at end of interval

```python
plt.step(x, y, where='post')
plt.title("Step Plot - Post")
plt.show()
```

where='mid'

Value change occurs at midpoint

2. linestyle Parameter

```python
plt.step(x, y, linestyle='--')   #Line style customize
plt.title("Step Plot with Dashed Line")
plt.show()
```

3.linewidth Parameter

```python
plt.step(x, y, linewidth=3)    #Professional visualization
plt.title("Thick Step Line")
plt.show()
```

4.marker Parameter

```python
plt.step(x, y, marker='o')          
plt.step(x, y, marker='o')           #To highlight Data points
plt.step(x, y, marker='o')      
plt.title("Step Plot with Markers")
plt.show()
```

5.color Parameter

```python
plt.step(x, y, color='blue')
plt.title("Colored Step Plot")
plt.show()
```

6.grid + axis control

```python
plt.step(x, y)
plt.grid(True)
plt.xlim(0,6)
plt.ylim(0,35)
plt.title("Step Plot with Grid and Axis Control")
plt.show()
```

Step Plot – Scenario 
---Real-World Scenario: Inventory Level Tracking

In inventory management systems, stock levels remain constant until a sale occurs.
When a sale happens, the stock quantity decreases suddenly.

This creates a piecewise constant behavior, where values change only at specific intervals.

A Step Plot is ideal for representing this type of data.

--Why Step Plot is Suitable Here:

Clearly shows interval-based changes

Represents constant values between events

Useful for business analytics and operational dashboards

Helps visualize stock drops over time

--This type of visualization is commonly used in:

Retail inventory systems

Electricity billing slabs

Subscription pricing models

CPU scheduling in operating systems

Network bandwidth allocation

Step 1: Import Libraries

```python
# Visualization library
import matplotlib.pyplot as plt

# Numerical data handling
import numpy as np
```

Step 2:Real Data

```python
# Time in hours
time = np.array([0, 2, 4, 6, 8, 10, 12])

# Stock levels (changes only when sale happens)
stock = np.array([100, 80, 80, 60, 60, 40, 20])
```

Step 3:Step Plot

```python
plt.figure(figsize=(8,5))

plt.step(
    time,
    stock,
    where='post',      # Stock updates after sale
    linestyle='-',
    linewidth=3,
    marker='o',
    color='blue'
)

plt.xlabel("Time (Hours)")
plt.ylabel("Stock Quantity")
plt.title("Inventory Level Tracking using Step Plot")

plt.grid(True)

plt.xlim(0, 13)
plt.ylim(0, 110)

plt.show()
```
