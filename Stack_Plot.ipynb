Stack Plot in Matplotlib

A Stack Plot is used to visualize multiple datasets stacked on top of each other.
It shows both individual contribution and cumulative total over time.

Stack plots are commonly used in:
- Revenue analysis
- Resource allocation
- Market share comparison
- Performance tracking

In Matplotlib, Stack Plot is created using the `stackplot()` function.

Import Libraries & Sample Data

```python
import matplotlib.pyplot as plt
import numpy as np
```

Sample Data

```python
x = np.arange(1, 6)  # Time (e.g., Months)

product_A = [10, 15, 20, 25, 30]
product_B = [5, 10, 15, 20, 25]
product_C = [2, 5, 8, 10, 12]
```

1.x Parameter

Defines X-axis values (time or categories)

```python
plt.stackplot(x, product_A)
plt.title("Using x Parameter")
plt.show()
```

2. Multiple y Values

Each list represents a stacked layer.

Shows contribution of multiple datasets.

```python
plt.stackplot(x, product_A, product_B, product_C)
plt.title("Using Multiple Y Values")
plt.show()
```

3.labels

Adds legend entries.

Important to identify each stacked component.

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    labels=["Product A", "Product B", "Product C"]
)

plt.legend()
plt.title("Using labels Parameter")
plt.show()
```

4. colors

Manually assigns color to each layer.


```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    colors=["skyblue", "lightgreen", "salmon"]
)

plt.title("Using colors Parameter")
plt.show()
```

5.alpha

Controls transparency (0–1).

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    alpha=0.6
)

plt.title("Using alpha Parameter")
plt.show()
```

6.baseline

Controls stacking alignment style.

Why Used:

'zero' → Default stacking

'sym' → Centered stack

'wiggle' → Streamgraph style

'weighted_wiggle' → Improved streamgraph balance

Used for advanced visualization like streamgraphs.
    

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    baseline='wiggle'   # Options: 'zero', 'sym', 'wiggle', 'weighted_wiggle'
)

plt.title("Using baseline Parameter")
plt.show()
```

7. hatch
   
Adds pattern inside layers.

Useful in grayscale printing or reports.

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    hatch='//'
)

plt.title("Using hatch Parameter")
plt.show()
```

8. edgecolor

Sets border color of stacked areas.

Improves visual separation between layers.

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    edgecolor='black'
    
)

plt.title("Using edgecolor Parameter")
plt.show()
```

9. linewidth
    
Controls thickness of borders.

Highlights layer boundaries clearly.
    

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    edgecolor='black',
    linewidth=2
)

plt.title("Using linewidth Parameter")
plt.show()
```

10.zorder

Controls layer drawing order.
    
Ensures correct stacking when combining multiple plots.

```python
plt.stackplot(
    x,
    product_A,
    product_B,
    product_C,
    zorder=3
)

plt.title("Using zorder Parameter")
plt.show()
```

11. data

Used when your data is stored inside a dictionary or pandas DataFrame.
    

```python
import matplotlib.pyplot as plt

data = {
    'years': [2019, 2020, 2021, 2022],
    'sales_a': [20, 25, 30, 35],
    'sales_b': [15, 18, 22, 28]
}

plt.stackplot('years', 'sales_a', 'sales_b',
              data=data,
              labels=['Product A', 'Product B'])

plt.legend()
plt.show()
```

Real World Scenario – Company Revenue Contribution Analysis

Scenario:

A company wants to analyze how different product categories contributed to total revenue over 6 years.

Categories:

Mobile Sales 📱

Laptop Sales 💻

Accessories Sales 🎧

The management wants to:

See total revenue growth

Understand individual product contribution

Observe trend shifts over time

Present it visually in board meetings

 Best visualization: Stack Plot

Step 1: Import Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
```

Step 2: Create Real-World Data

```python
# X-axis data (Years)
years = [2018, 2019, 2020, 2021, 2022, 2023]

# Revenue data (in millions)
mobile_sales = [20, 25, 30, 35, 40, 45]
laptop_sales = [15, 18, 22, 28, 30, 35]
accessories_sales = [5, 8, 10, 12, 15, 18]
```

Step 3: stack Plot Covering All Parameters

```python
# Creating Stack Plot
plt.figure(figsize=(10,6))  # Adjust figure size

plt.stackplot(
    years,                        # X-axis values
    mobile_sales,                 # Y1 data
    laptop_sales,                 # Y2 data
    accessories_sales,            # Y3 data
    
    labels=['Mobile', 'Laptop', 'Accessories'],  # Legend labels
    colors=['skyblue', 'orange', 'green'],       # Area colors
    alpha=0.8,                    # Transparency level
    edgecolor='black',            # Border color
    linewidth=1.5,                # Border thickness
    linestyle='-',                # Border style
    hatch='//',                   # Pattern style
    baseline='zero'               # Default stacking from zero
)

# Adding Title and Labels
plt.title("Company Revenue Contribution (2018-2023)")
plt.xlabel("Year")
plt.ylabel("Revenue (in Millions)")

# Display legend
plt.legend(loc='upper left')

# Grid for better readability
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.show()
```

-- Why Stack Plot is Used Here?

See total revenue growth -	Shows cumulative area

Compare categories       - 	Clear color separation

Identify trend shifts	 -   Visual slope change

Board presentation	     -  Professional look
